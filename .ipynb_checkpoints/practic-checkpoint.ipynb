```python
import pandas as pd
import numpy as np
df = pd.read_excel(r"C:\Users\dell\Downloads\Financial_Sample.xlsx")
```

```python
df.info()
```

```python

for i in df.select_dtypes(include="str").columns:
    df[i] = df[i].astype("object")
```

```python
df.info()
```

```python
# 'str' ya string dtype columns ko explicitly 'object' dtype me convert karein
for col in df.columns:
    if df[col].dtype == 'str' or 'string' in str(df[col].dtype):
        df[col] = df[col].astype('object')
```

```python
for col in df.columns:
    if df[col].isnull().sum()>0:
        print(col,'\t',df[col].isnull().sum(),'\t',df[col].dtype)
```

```python
# fill missing values numerical with mean(average) and categorical/text with mode(most_frequent)
for col in df.columns:
    if df[col].isnull().sum()>0:
        if df[col].dtype in [object , str]:
            print(col ,'\t', df[col].isnull().sum() ,'\t', df[col].dtype,'\t Mode : ',df[col].mode()[0])
            df[col] = df[col].fillna( df[col].mode()[0] )
        else:
            print(col ,'\t', df[col].isnull().sum() ,'\t', df[col].dtype , '\t Mean : ' , round(df[col].mean(),2))
            df[col] = df[col].fillna( round(df[col].mean(),2) )
```
