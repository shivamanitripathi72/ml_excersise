```python
keys = ["a", "b", "c"]
values = [10, 20, 30]
res = {keys[i]: values[i] for i in range(len(keys))}
print(res)
```

```python
keys = ["name", "age", "city"]
values = ["Rahul", 25, "Delhi"]
res = {}
for i in range(len(keys)):
    res[keys[i]] = values[i]
print(res)
```

```python
keys = ["a", "b", "c"]
values = [1, 2, 3]
# Dictionary comprehension
result = {k: v for k, v in zip(keys, values)}
print(result)
```

```python
keys = ["name", "age", "city"]
values = ["Rahul", 25, "Delhi"]

# zip() aur dict() se dictionary banana
result = dict(zip(keys, values))
print(result)
```

```python
def add(a,b):
    return a * b
print(add(4,5))
```

```python
def calculate(a, b):
    # Instruction 1: Addition karna
    sum_val = a + b
    
    # Instruction 2: Multiplication karna
    prod_val = a * b
    
    # Instruction 3: Screen par print karna
    print(f"Sum: {sum_val}")
    
    # Instruction 4: Result wapas bhejna
    return prod_val

# Function call
res = calculate(4, 5)
print("Product:", res)
```

```python
def get_length(my_list):
    count = 0
    for i in (my_list):
        count += 1
    return count
num = [1,2,3,4,5,6,7,8]
get_length(num)
```

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df  = sns.load_dataset('iris')
df.head()
```

```python
x = df.drop(columns=['species'])
y = df['species']
```

```python
from sklearn.preprocessing import StandardScaler
x_scalled = StandardScaler().fit_transform(x)
x_scalled[:3]
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(x_scalled)
X_pca[:4]
```

```python
df_pca = pd.DataFrame(
    X_pca , columns=['PCA1','PCA2']
)
```

```python
df_pca.head(2)
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
lda = LinearDiscriminantAnalysis( n_components=2 )
X_lda = lda.fit_transform(x_scalled , y) 
X_lda.shape
```

```python
# LDA ke results ka DataFrame banana
df_lda = pd.DataFrame(X_lda, columns=['LDA1', 'LDA2'])
df_lda['species'] = y.values

# Scatter plot banana
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df_lda, x='LDA1', y='LDA2', hue='species', palette='Set1', s=70)
plt.title('LDA Projection of Iris Dataset')
plt.xlabel('LDA 1')
plt.ylabel('LDA 2')
plt.grid(True)
plt.show()
```
