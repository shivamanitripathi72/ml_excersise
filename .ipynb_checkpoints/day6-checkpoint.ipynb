```python
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
import seaborn as sns
import matplotlib.pyplot as plt
kmeans = KMeans( n_clusters = 2 )
```

```python
df  = sns.load_dataset('titanic').bfill()
df.head()
```

```python
x = df['fare']
x
```

```python
y  = df['age']
y
```

```python
g = np.column_stack((x,y))
labels = kmeans.fit_predict(g)
```

```python
labels
```

```python
plt.figure( figsize=(6,3) )
sns.scatterplot( x= x,y = y , c=df['survived'])
plt.show()
```

```python
plt.figure( figsize=(6,3) )
sns.scatterplot( x= x,y = y , c =labels)
plt.show()
```
