```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
print(sns.get_dataset_names())
```

```python
df = sns.load_dataset('attention')
df.head()
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df.attention.unique()
```

```python
x = df['attention']
df = df.drop(columns=(['attention']))
df.head()
```

```python
df.head()
```

```python
from sklearn.cluster import KMeans

# 1. KMeans initialize karein
kmeans = KMeans(n_clusters=2)

# 2. Fit karke labels assign karein
df['cluster'] = kmeans.fit_predict(df)
```

```python
df.head()
```

```python
plt.figure(figsize=(8, 4))

# Scatter plot: x-axis par 'timepoint' aur y-axis par 'subject'
sns.scatterplot(
    data=df, 
    x='score', 
    y= 'solutions', 
    hue='cluster', 
    palette='viridis', 
    s=70, 
    alpha=0.8
)

plt.title('K-Means Clusters Visualization')
plt.xlabel('score')
plt.ylabel('solutions (Encoded)')
plt.grid(True, linestyle='--', alpha=0.9)
plt.show()

plt.figure(figsize=(8, 4))
sns.scatterplot(
    data=df, 
    x='score', 
    y= 'solutions', 
    hue= x, 
    palette='viridis', 
    s=70, 
    alpha=0.8
)

plt.title('K-Means Clusters Visualization')
plt.xlabel('score')
plt.ylabel('solutions (Encoded)')
plt.grid(True, linestyle='--', alpha=0.9)
plt.show()
```
