```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# Step 1: Customer + Date + Time ke basis par Category ki list banayein

# Step 2: TransactionEncoder chalayein
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df_encoded = pd.DataFrame(te_ary, columns=te.columns_)

# Step 3: Frequent itemsets nikalein
frequent_itemsets = apriori(df_encoded, min_support=0.01, use_colnames=True)

# Step 4: Association rules generate karein
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.0)
rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].sort_values('lift', ascending=False)
```

```python
# 1. Group unique categories per (Customer_ID, Date, Time) using a dictionary
transaction_dict = {}

for _, row in df.iterrows():
    key = (row["Customer_ID"], row["Date"], row["Time"])
    category = row["Category"]

    if key not in transaction_dict:
        transaction_dict[key] = set()

    transaction_dict[key].add(category)

# 2. Convert the dictionary values into a list of lists
transactions = [list(categories) for categories in transaction_dict.values()]
```

```python
transactions_dict = {}

for row in df[['Customer_ID', 'Date', 'Time', 'Category']].itertuples(index=False):
    key = (row.Customer_ID, row.Date, row.Time)
    transactions_dict.setdefault(key, set()).add(row.Category)

transactions = list(transactions_dict.values())
transactions = [list(items) for items in transactions]
```
